```python
"""
Add Binary

Given two binary strings a and b, return their sum as a binary string.

Example 1:

Input: a = "11", b = "1"
Output: "100"
Example 2:

Input: a = "1010", b = "1011"
Output: "10101"
"""

def addBinary(a,b):
    return bin(int(a, 2) + int(b, 2))[2:] # [2:] is used to remove 0b at the beginning of the string

a = "1010"
b = "1011"

print(addBinary(a,b))
```

```python
def addBinaryManual2(a,b):
    i,j = len(a) - 1, len(b) - 1
    carry = 0
    res = []
    while i >= 0 or j >= 0 or carry:
        if i >= 0:
            carry += int(a[i])
            i -= 1
        if j >= 0:
            carry += int(b[j])
            j -= 1
        res.append(str(carry % 2))
        carry = carry // 2
    return "".join(res[::-1])

a = "11"
b = "1"
print(addBinaryManual2(a,b))
```
