```python
# sum of nested list values
data = [[1,2,3],[4,5],[6,7,8,9]]

result = sum([item for sublist in data for item in sublist])
print(result)
```

```python
# Count elements in nested lists
result = sum(len(sublist) for sublist in data)
print(result)
```

```python
# Extract all names from list of dicts
users = [
    {"id":1,"name":"A"},
    {"id":2,"name":"B"},
    {"id":3,"name":"C"}
]

result = []
for sub_dict in users:
    result.append(sub_dict['name'])
print(result)
```

```python
# Students scoring above threshold
threshold = 60
students = [
    {"name":"A","marks":[70,80,65]},
    {"name":"B","marks":[40,50,45]},
    {"name":"C","marks":[90,88,92]}
]

final_list = []
for s in students:
    if s['marks']:
        average = sum(s['marks'])/len(s['marks'])
        if average >= threshold:
            final_list.append(s['name'])

print(final_list)
```

```python
# Flatten nested dictionary values
data = {
    "a":[1,2],
    "b":[3,4,5],
    "c":[6]
}
print(data.keys())
print(data.values())
result_set = []
result_set.extend([*data])
for i in data.values():
    result_set.extend(j for j in i)
print(result_set)
```

```python
"""Python function to flatten nested iterable"""
def flatten_values(obj):
    if isinstance(obj, dict):
        for value in obj.values():
            yield from flatten_values(value)
    elif isinstance(obj, list):
        for item in obj:
            yield from flatten_values(item)
    else:
        yield obj
```

```python
# flatten this
data = {
    "a":[1,2],
    "b":[3,[4,5]],
    "c":[6,[7,[8,9]]]
}
# expected output [1,2,3,4,5,6,7,8,9]

result_set = []

for i in data.values():
    result_set.extend(flatten_values(i))
print(f"result: {result_set}")
```

```python
# realistic challenge
logs = [
 {"user":"A","page":"home","time":3,"device":"mobile"},
 {"user":"B","page":"cart","time":5,"device":"web"},
 {"user":"A","page":"prod","time":2,"device":"web"},
 {"user":"A","page":"cart","time":8,"device":"mobile"},
 {"user":"B","page":"home","time":1,"device":"mobile"}
]

# o/p
"""{
 "A":{
      "mobile":["home","cart"],
      "web":["prod"]
    },
 "B":{
      "mobile":["home"],
      "web":["cart"]
    }
}"""

from collections import defaultdict

groups = defaultdict(lambda: defaultdict(list))

for log in logs:
    groups[log['user']][log['device']].append(log)

#print(groups)

# sort each list by time and keep only pages
result = {}

for user, device in groups.items():
    result[user] = {}
    for device, entries in device.items():
        entries.sort(key=lambda x:x['time'])
        result[user][device] = [e['page'] for e in entries]

print(result)
```
