```python
"""
Minimum cost to achieve target average capacity

You are given n servers. Each server has:
1) An initial computation capacity given by computationCap[i]
2) A cost per unit increase in capacity given by cost[i]
3) A maximum capacity limit given by maxCap

You can increase the capacity of any server any number of times, but:
> Each unit increase in capacity for server i costs cost[i]
> The capacity of server i cannot exceed maxCap

DETERMINE the minimum total cost required to ensure that the average computation capacity of all servers is at least avgCap
"""

# Example
# computationCap = [2,3,1]
# cost = [2,1,3]
# maxCap = 5
# avgCap = 4

# Output = 11

"""
Explanation:
Current total capacity =  2 + 3 + 1 = 6
Required total = 3 * 4 = 12
Deficit = 6
To minimize cost:
> Increase the server with lowest cost first
> Apply upgrades optimally unitl total reaches 12
"""
```

```python
def min_cost_to_reach_avg(computationCap, cost, maxCap, avgCap):
    n = len(computationCap)

    current_total = sum(computationCap)
    required_total = n * avgCap

    # check if already satisfies condition
    if current_total >= required_total:
        return 0

    deficit = required_total - current_total

    servers = [] # create a list of (cost per unit, max possible increase)
    
    for i in range(n):
        max_increase = maxCap - computationCap[i]
        if max_increase > 0:
            servers.append((cost[i], max_increase))

    servers.sort() # sort by cost per unit (cheapest first)
    total_cost = 0
    for c, max_inc in servers:
        if deficit <= 0:
            break
        take = min(deficit, max_inc)
        total_cost += take * c
        deficit -= take

    # if still deficit remains, its impossible
    if deficit > 0:
        return -1
    return total_cost
```

```python
computationCap = [2,3,1]
cost = [2,1,3]
maxCap = 5
avgCap = 4

print(min_cost_to_reach_avg(computationCap, cost, maxCap, avgCap))
```
