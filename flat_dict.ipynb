```python
def flatten_generic_list(data, parent_key='', sep='.'):
    flat = {}
    if isinstance(data, dict):
        for k, v in data.items():
            new_key = f"{parent_key}{sep}{k}" if parent_key else k
            flat.update(flatten_generic_list(v, new_key, sep))
    elif isinstance(data, list):
        for i, value in enumerate(data):
            new_key = f"{parent_key}{sep}{i}"
            flat.update(flatten_generic_list(value, new_key, sep))

    else:
        flat[parent_key] = data
    return flat
```

```python
data = {
    "user": {
        "name": "Arnab",
        "skills": ["python", "aws"],
        "projects": {
            "p1": {"status": "done"},
            "p2": {"status": "wip"}
        }
    },
    "age": 25
}
```

```python
result = flatten_generic_list(data)
for k, v in result.items():
    print(k,":",v)
```
